```python
import scrapy
```

```python
class Crawler(scrapy.Spider):
    name = "docu_spider"
    allowed_domains = ["chotot.com"]
    
    start_urls = ["https://www.chotot.com/mua-ban-laptop-ha-noi?event_source=ct_home_highlight_category"]
    
    def parse(self, response):
        detail_links = response.css("#__next > div > div.container.c2c-container.ct-listing > div.styles_listViewWrapper__gETtz > div > main > div.no-padding.col-md-12.ct-m-x-876 > div.list-view > div > div > ul > div:nth-child(1) > div > li > div > a").getall()
        
        for link in detail_links:
            yield response.follow(link, callback=self.parse_detail)
            
        next_page = response.css("#__next > div > div.container.c2c-container.ct-listing > div.styles_listViewWrapper__gETtz > div > main > div.no-padding.col-md-12.ct-m-x-876 > div.col-xs-12.no-padding > div > div > div:nth-child(11) > a").get()
        if next_page is not None:
            yield response.follow(next_page, callback=self.parse)
            
    def parse_detail(self, response):
        title = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.c16nxtni > div > div.ctcdcee > div:nth-child(1) > div:nth-child(1) > div.r9vw5if > div > div > h1").get()
        price = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.c16nxtni > div > div.ctcdcee > div:nth-child(1) > div:nth-child(3) > b").get()
        description = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s67mqju > div > div.c2yxoiv > p").get()
        brand = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(1) > div.p1vpox21").get()
        line = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(2) > div.p1vpox21").get()
        device_condition = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(3) > div.p1vpox21").get()
        warranty_policy = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(4) > div.p1vpox21").get()
        screen_size = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(5) > div.p1vpox21").get()
        cpu = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(6) > div.p1vpox21").get()
        ram = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(7) > div.p1vpox21").get()
        graphic_card = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(8) > div.p1vpox21").get()
        hard_drive = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(9) > div.p1vpox21").get()
        hard_drive_type = response.css("#__next > div > div.c1qees99 > div.c4vh6ke > div.r9vw5if > div > div.left.s7721ok.r9vw5if > div.s1qfavim > div.c189eou8 > div > div:nth-child(11) > div.p1vpox21").get()
        
        if description:
            description = description.strip()
            
        results = {
            "URL": response.url,
            "Title": title,
            "Price": price,
            "Description": description,
            "Brand": brand,
            "Line": line,
            "Device Condition": device_condition,
            "Warranty Policy": warranty_policy,
            "Screen Size": screen_size,
            "CPU": cpu,
            "RAM": ram,
            "Graphic Card": graphic_card,
            "Hard Drive": hard_drive,
            "Hard Drive Type": hard_drive_type
        }
            
        return results
        '''yield {
            "URL": response.url,
            "Title": title,
            "Price": price,
            "Description": description,
            "Brand": brand,
            "Line": line,
            "Device Condition": device_condition,
            "Warranty Policy": warranty_policy,
            "Screen Size": screen_size,
            "CPU": cpu,
            "RAM": ram,
            "Graphic Card": graphic_card,
            "Hard Drive": hard_drive,
            "Hard Drive Type": hard_drive_type
        }'''
```
